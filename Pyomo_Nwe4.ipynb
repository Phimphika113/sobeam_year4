```python
import calendar
from datetime import datetime

# กำหนดเดือนที่ต้องการ 12 เดือน
month = "June"  # เปลี่ยนเดือนที่ต้องการใช้งานที่นี่
year = 2024  # กำหนดปี ค.ศ.

# หาจำนวนวันในเดือนที่เลือกโดยใช้ calendar module
D = calendar.monthrange(year, datetime.strptime(month, '%B').month)[1]

# สร้างตัวแปรสำหรับเก็บวันและวันในสัปดาห์
days_with_weekdays = []
# สร้างพจนานุกรมสำหรับเก็บจำนวนวันในแต่ละเดือน
days_in_month = {}
# เพิ่มค่า D ลงในพจนานุกรม days_in_month
days_in_month[month] = D

# หาวันหยุดเสาร์-อาทิตย์
weekends = [
    day
    for day in range(1, D + 1)
    if datetime.strptime(f"{year}-{month}-{day:02d}", "%Y-%B-%d").strftime("%A") in ["Saturday", "Sunday"]
]
weekends_count = len(weekends)

# ใช้ datetime เพื่อหาวันในสัปดาห์
for day in range(1, D + 1):  # ใช้ range() ตรงนี้
    date_str = f"{year}-{month}-{day:02d}"  # สร้าง string ของวัน
    date_obj = datetime.strptime(date_str, "%Y-%B-%d")  # แปลงเป็น datetime object
    weekday = date_obj.strftime("%A")  # ค้นหาวันในสัปดาห์
    days_with_weekdays.append((day, weekday))  # เพิ่มในลิสต์

# แสดงผล
for day, weekday in days_with_weekdays:
    print(f"Day {day}: {weekday}")
```

```python
# ข้อมูลพยาบาลต่างๆ
Mij = {(i, j): 0 for i in nurses for j in days}  # พยาบาลขอหยุดทำงาน # เริ่มต้นด้วยค่า 0
Vij = {(i, j): 0 for i in nurses for j in days}
Bij = {(i, j): 0 for i in nurses for j in days}  # พยาบาลลาพักร้อน
di = {i: 0 for i in nurses}  # พยาบาลที่มีโรคประจำตัว
ages = {i: 0 for i in nurses}  # อายุของพยาบาล
work_experience = {i: 0 for i in nurses}  # อายุการทำงานของพยาบาล
```

```python
# กำหนดวันหยุดในรูปแบบลิสต์สำหรับแต่ละพยาบาล
days_off = {
    1: [7,8,9,10,16,25,26,27],
    2: [1,2,3,12,15,23,27],
    3: [10,18,19,20,26],
    4: [2,12,17,27,28,29,30],
    5: [1,2,3,9,10,11,21,22],
    6: [8,11,21,26,27,28,29,30],
    7: [1,2,9,14,15,16,18,19,20],
    8: [8,9,10,11,20,25],
    9: [3,4,13,18,19,20,29],
    10: [6,14,22,23,24,25],
    11: [3,4,5,6,14,18,19,20,28],
    12: [7,8,9,13,14,15,24],
    13: [5,11,15,16,17,22,23,24,30],
    14: [1,2,3,4,13,21,25,],
    15: [6,12,13,22,28,29,30],
    16: [1,2,7,15,16,25,30],
    17: [1,9,17,22,23],
}

# ใช้ลูปเพื่อกำหนดวันหยุด
for nurse, off_days in days_off.items():
    for day in off_days:
        Mij[(nurse, day)] = 1
```

```python
#แสดงจำนวนวันที่พยาบาลต้องทำงานในแต่ละเดือน
alpha = min_work_days_per_month
```

##5. สร้างโมเดล

```python
# กำหนดพารามิเตอร์ใน Pyomo
model.M = Param(nurses, days, initialize=Mij, within=Binary)
model.V = Param(nurses,days, initialize=Vij, within=Binary)
model.B = Param(nurses, days, initialize=Bij, within=Binary)
model.d = Param(nurses, initialize=di, within=Binary)
model.ages = Param(nurses, initialize=ages, within=NonNegativeIntegers)
model.work_experience = Param(nurses, initialize=work_experience, within=NonNegativeIntegers)
```

```python
# กำหนดตัวแปร NT และ OT (เวรปกติและเวรล่วงเวลา)
model.NT = Var(model.nurses, within=NonNegativeIntegers)  # เวรปกติ
model.OT = Var(model.nurses, within=NonNegativeIntegers)  # เวรล่วงเวลา
```

```python
# กำหนดตัวแปรเสริมสำหรับแทนค่าค่าสัมบูรณ์
model.pos_NT = Var(model.nurses, within=NonNegativeReals)
model.neg_NT = Var(model.nurses, within=NonNegativeReals)
model.pos_OT = Var(model.nurses, within=NonNegativeReals)
model.neg_OT = Var(model.nurses, within=NonNegativeReals)
```

```python
# กำหนดข้อจำกัดเพื่อแทนที่ฟังก์ชันค่าค่าสัมบูรณ์
def abs_NT_constraint(model, i):
    return model.NT[i] - alpha == model.pos_NT[i] - model.neg_NT[i]
model.abs_NT_constraint = Constraint(model.nurses, rule=abs_NT_constraint)

def abs_OT_constraint(model, i):
    return model.OT[i] - beta == model.pos_OT[i] - model.neg_OT[i]
model.abs_OT_constraint = Constraint(model.nurses, rule=abs_OT_constraint)


```

##สมการวัตถุประสงค์

```python
# ฟังก์ชันวัตถุประสงค์ SAD แบบเชิงเส้น
def objective_rule(model):
    return sum(model.pos_NT[i] + model.neg_NT[i] for i in model.nurses) + \
           sum(model.pos_OT[i] + model.neg_OT[i] for i in model.nurses)

model.objective = Objective(rule=objective_rule, sense=minimize)
```

##สมการเงื่อนไข/ข้อจำกัด

```python
# ข้อจำกัดพยาบาลที่มีอายุ 50 ปีขึ้นไปห้ามทำงานดึก
def constraint_3_5_rule(model, i, j):
    if ages[i] >= 50:
        return model.X[i, j, 3] == 0
    return Constraint.Skip
model.constraint_3_5 = Constraint(nurses, days, rule=constraint_3_5_rule)
```

```python
# ข้อจำกัดพยาบาลที่มีอายุ 50 ปีขึ้นไปต้องทำงานเช้ามากกว่าบ่าย
def age_50_plus_morning_shifts_rule(model, i):
    if ages[i] >= 50:
        return sum(model.X[i, j, 1] for j in days) >= sum(model.X[i, j, 2] for j in days)
    return Constraint.Skip
model.age_50_plus_morning_shifts = Constraint(nurses, rule=age_50_plus_morning_shifts_rule)

```

```python
# เพิ่มข้อจำกัดในการทำงานของพยาบาล วันหยุด (Mij)
def nurse_off_days_rule(model, i, j):
    return sum(model.X[i, j, k] for k in model.shifts) <= (1 - Mij[(i, j)]) * len(model.shifts)

# กำหนดข้อจำกัดให้กับโมเดล
model.nurse_off_days_constraint = Constraint(model.nurses, model.days, rule=nurse_off_days_rule)

```

```python
# ข้อจำกัด: พยาบาลต้องไม่ทำงานในวันที่ขอลาพักผ่อน (Vij)
def vacation_request_rule(model, i, j):
    return sum(model.X[i, j, k] for k in model.shifts) <= (1 - Vij[(i, j)]) * len(model.shifts)

model.vacation_request_constraint = Constraint(model.nurses, model.days, rule=vacation_request_rule)
```

```python
# ข้อจำกัดห้ามทำงานผลัดเช้าติดกับผลัดบ่ายในวันเดียวกัน
def no_morning_and_afternoon_same_day_rule(model, i, j):
    return model.X[i, j, 1] + model.X[i, j, 2] <= 1

# เพิ่มข้อจำกัดในโมเดล
model.no_morning_and_afternoon_same_day_constraint = Constraint(model.nurses, model.days, rule=no_morning_and_afternoon_same_day_rule)
```

```python
# ข้อจำกัดห้ามทำงานดึกแล้วเช้าในวันถัดไป
def no_night_to_morning_shifts_rule(model, i, j):
    if j < D:
        return model.X[i, j, 3] + model.X[i, j+1, 1] <= 1
    return Constraint.Skip
model.no_night_to_morning_shifts_constraint = Constraint(nurses, range(1, D), rule=no_night_to_morning_shifts_rule)
```
